## 结果报告

## 回归模型
- R2拟合程度：因变量42%的异变可以由模型解释，拟合程度中等。
- 模型p-value: 0.0000000000139 < 0.05，模型显著
- 变量回归系数：
  - 1. 参与线上营销活动的意愿：0.36 (p=0.00)
  - 2. 在网红身上的自我反射：0.08 (p=0.34)
  - 3. 对网红的信任：0.13 (p=0.07)
研究显示，仅用户自身对于线上营销活动的参与意愿显著影响用户对于网红营销的参与度，并且也显示该变量对于因变量的回归系数是最高的，意味着每一单位的该变量的异动在控制其他变量的情况下会导致因变量0.36个单位的异动。另外，用户对网红的信任程度对因变量产生影响的回归系数为0.13，且p值达到了0.07。虽然未达到显著结论的阈值，但仍接近阈值，因此该变量对于因变量的影响仍不可忽视。但由于其对于因变量的回归系数较低，因此该变量对因变量的贡献度较低。
## 模型评价
残差正态性
- Omnibus / JB 检验：p值均大于0.05，无法拒绝残差正态分布的原假设
残差自相关性
- Durbin-Watson 检验：该系数达到1.90, 约等于2且坐落于[1.5,2.5]的区间。因此残差无自相关
多重共线性
- VIF检验：VIF值均小于5，因此自变量无自相关性

## 阶级聚类分析

用户被划分为4个簇：

1. 网络沉浸用户：男性为主，高受教程度，低收入，热爱网络的一切。
2. 现实优先用户：女性为主，低受教程度，低收入，不爱关注网络。
3. 网络冷感用户：女性为主，高受教程度，低收入，对网络有偏见，但喜新鲜事物，不喜社交。
4. 理性稳定用户：男性为主，高受教程度，高收入，对网络不敏感，且不喜社交。
